```python
# Exp 13: Handle missing values and inappropriate data in a given data set (Hotel_Dataset.csv)
# Colab: upload Hotel_Dataset.csv via the Files panel (left sidebar), then paste each "# %%" block into its own cell.

# %% Cell 1 - read the CSV
import numpy as np
import pandas as pd
df = pd.read_csv("/content/Hotel_Dataset.csv")
df

# %% Cell 2 - find duplicate rows (row 9 is repeated)
df.duplicated()

# %% Cell 3 - info
df.info()

# %% Cell 4 - remove duplicates
df.drop_duplicates(inplace=True)
df

# %% Cell 5 - number of rows
len(df)

# %% Cell 6 - reset the index
index = np.array(list(range(0, len(df))))
df.set_index(index, inplace=True)
index

# %% Cell 7
df

# %% Cell 8 - drop the duplicate column (CSV has Age_Group twice, pandas names the 2nd 'Age_Group.1')
df.drop(['Age_Group.1'], axis=1, inplace=True)
df

# %% Cell 9 - inappropriate (negative) values -> NaN
df.loc[df.CustomerID < 0, 'CustomerID'] = np.nan
df.loc[df.Bill < 0, 'Bill'] = np.nan
df.loc[df.EstimatedSalary < 0, 'EstimatedSalary'] = np.nan


# %% Cell 10 - NoOfPax must be between 1 and 20
df.loc[(df['NoOfPax'] < 1) | (df['NoOfPax'] > 20), 'NoOfPax'] = np.nan


# %% Cell 11 - unique values of the categorical columns
df.Age_Group.unique()

# %% Cell 12
df.Hotel.unique()

# %% Cell 13 - fix the wrong hotel spelling
df['Hotel'] = df['Hotel'].replace(['Ibys'], 'Ibis')

# %% Cell 14
df.FoodPreference.unique()

# %% Cell 15 - make food preference consistent
df['FoodPreference'] = df['FoodPreference'].replace(['Vegetarian', 'veg'], 'Veg')
df['FoodPreference'] = df['FoodPreference'].replace(['non-Veg'], 'Non-Veg')

# %% Cell 16 - fill missing values
# continuous numerical -> mean, discrete numerical -> median, categorical -> mode
df['EstimatedSalary'] = df['EstimatedSalary'].fillna(round(df.EstimatedSalary.mean()))
df['NoOfPax'] = df['NoOfPax'].fillna(round(df.NoOfPax.median()))
df['Rating(1-5)'] = df['Rating(1-5)'].fillna(round(df['Rating(1-5)'].median()))
df['Bill'] = df['Bill'].fillna(round(df.Bill.mean()))
df
```
