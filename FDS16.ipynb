```python
# Exp 15: EDA - Quantitative and Qualitative analysis (seaborn 'tips' dataset)
# Colab: no CSV upload needed. Copy each "# %%" block into its own cell and run in order.

# %% Cell 1 - imports
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

# %% Cell 2 - load the built-in dataset (needs internet, Colab has it)
tips = sns.load_dataset('tips')
tips.head()

# %% Cell 3 - distribution of total_bill
sns.displot(tips.total_bill, kde=True)

# %% Cell 4
sns.displot(tips.total_bill, kde=False)

# %% Cell 5 - tip vs total_bill
sns.jointplot(x=tips.tip, y=tips.total_bill)

# %% Cell 6
sns.jointplot(x=tips.tip, y=tips.total_bill, kind="reg")

# %% Cell 7
sns.jointplot(x=tips.tip, y=tips.total_bill, kind="hex")

# %% Cell 8 - pairplots
sns.pairplot(tips)

# %% Cell 9
tips.time.value_counts()

# %% Cell 10
sns.pairplot(tips, hue='time')

# %% Cell 11
sns.pairplot(tips, hue='day')

# %% Cell 12 - correlation heatmap
sns.heatmap(tips.corr(numeric_only=True), annot=True)

# %% Cell 13 - boxplots (outliers)
sns.boxplot(y=tips.total_bill)

# %% Cell 14
sns.boxplot(y=tips.tip)

# %% Cell 15 - qualitative: counts per category
sns.countplot(y=tips.day)

# %% Cell 16
sns.countplot(y=tips.sex)

# %% Cell 17
tips.sex.value_counts().plot(kind='pie')

# %% Cell 18
tips.sex.value_counts().plot(kind='bar')

# %% Cell 19 - dinner days only
sns.countplot(y=tips[tips.time == 'Dinner']['day'])
```
