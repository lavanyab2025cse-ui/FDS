```python
# Exp 11: Pandas library use cases in Data Science
# Colab: first upload 50_Startups.csv and employee.csv using the Files panel (left sidebar)
# Each "# %%" block below is one Colab cell.

# %% Cell 1 - imports
import numpy as np
import pandas as pd

# %% Cell 2 - DataFrame from a list
list = [[1, 'Smith', 50000], [2, 'Jones', 60000]]
df = pd.DataFrame(list)
df

# %% Cell 3 - name the columns
df.columns = ['Empd', 'Name', 'Salary']
df

# %% Cell 4 - info
df.info()

# %% Cell 5 - read 50_Startups.csv
df = pd.read_csv("/content/50_Startups.csv")
df.info()

# %% Cell 6 - head
df.head()

# %% Cell 7 - tail
df.tail()

# %% Cell 8 - read employee.csv
df = pd.read_csv("/content/employee.csv")
df.head()

# %% Cell 9 - tail and info
df.tail()

# %% Cell 10
df.info()

# %% Cell 11 - a column as a Series
df.salary

# %% Cell 12
type(df.salary)

# %% Cell 13 - mean
df.salary.mean()

# %% Cell 14 - median
df.salary.median()

# %% Cell 15 - mode
df.salary.mode()

# %% Cell 16 - variance
df.salary.var()

# %% Cell 17 - standard deviation
df.salary.std()

# %% Cell 18 - describe
df.describe()

# %% Cell 19 - describe all columns
df.describe(include='all')

# %% Cell 20 - column names
empCol = df.columns
empCol

# %% Cell 21 - DataFrame to NumPy array
emparray = df.values
emparray

# %% Cell 22 - NumPy array back to DataFrame
employee_DF = pd.DataFrame(emparray, columns=empCol)
employee_DF
```
